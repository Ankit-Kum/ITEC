```python
import numpy as np
```

```python
x = np.array([[1,0,0],[0,0,1],[1,0,1],[1,1,0]])
y = np.array([[0],[0],[0],[1]])
```

```python
print(x)
```

```python
print(y)
```

```python
## Assign Weights
np.random.seed(10)
weights = np.random.random((3,1))
weights
```

```python
### Weighted Sum
sum = np.dot(x,weights) + 0.02   ## (0.02 = bias)
sum
```

```python
### Activation Function
yp = 1 / (1 + np.exp(-sum))
yp
```

```python
yp.round()
```

```python
#a = np.where(yp>=0.7,1,0)
#a
```

```python
### Gradient Descent & Back Propagation (Optimize Model)
sum = np.dot(x,weights) + 0.02
yp = 1 / (1 + np.exp(-sum))
error = y - yp
change_in_weight = error * yp * (1-yp)
weights = weights + np.dot(x.T,change_in_weight)
error
```

```python
x.T.shape
```

```python
change_in_weight.shape
```

```python
### Training
for i in range(1000):
  sum = np.dot(x,weights) + 0.02
  yp = 1 / (1 + np.exp(-sum))
  error = y - yp
  change_in_weight = error * yp * (1-yp)
  weights = weights + np.dot(x.T,change_in_weight)
```
