```python
import numpy as np
```

```python
x = np.array([[1,0,0],[0,0,1],[1,0,1],[1,1,0]])
y = np.array([[0],[0],[0],[1]])
```

```python
print(x)
```

```python
print(y)
```

```python
## Assign Weights
np.random.seed(10)
weights = np.random.random((3,1))
weights
```

```python
### Weighted Sum
sum = np.dot(x,weights) + 0.02   ## (0.02 = bias)
sum
```

```python
### Activation Function
yp = 1 / (1 + np.exp(-sum))
yp
```

```python
yp.round()
```
