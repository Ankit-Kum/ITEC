```python
import nltk
```

```python
data = """Natural language processing (NLP) is an interdisciplinary subfield of linguistics,
computer science, and artificial intelligence concerned with the interactions between computers
and human language, in particular how to program computers to process and analyze large amounts of
natural language data. The goal is a computer capable of "understanding" the contents of documents,
including the contextual nuances of the language within them. The technology can then accurately extract
information and insights contained in the documents as well as categorize and organize the documents themselves"""
```

```python
print(data)
```

```python
nltk.download('punkt')
```

```python
## Tokenization
words = nltk.word_tokenize(data)
print(words)
```

```python
len(words)
```

```python
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
```

```python
nltk.download('stopwords')
```

```python
stopwords.words("english")
```

```python
sentences = nltk.sent_tokenize(data)
sentences
```

```python
len(sentences)
```

```python
### Stemming + Remove Stopwords
sentences = nltk.sent_tokenize(data)
stemmer = PorterStemmer()
for i in range(len(sentences)):
  words = nltk.word_tokenize(sentences[i])
  final_words = [stemmer.stem(x) for x in words if x.lower() not in set(stopwords.words("english"))]
  sentences[i] = " ".join(final_words)
```

```python
sentences
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('wordnet')
```

```python
### Stemming + Remove Stopwords
sentences = nltk.sent_tokenize(data)
lemma = WordNetLemmatizer()
for i in range(len(sentences)):
  words = nltk.word_tokenize(sentences[i])
  final_words = [lemma.lemmatize(x) for x in words if x.lower() not in set(stopwords.words("english"))]
  sentences[i] = " ".join(final_words)
```

```python
sentences
```

```python
### Bag Of Words
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features = 1500)
features = cv.fit_transform(sentences).toarray()
features
```

```python
### TF*IDF
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tf = TfidfVectorizer(max_features = 1500)
features = tf.fit_transform(sentences).toarray()
features
```

```python
### Named Entity Recognition
```

```python
import spacy
nlp = spacy.load('en_core_web_sm')
```

```python
text = nlp("Google and Amazon is looking for a Data Science Engineer having an experience of 5 years and offering $20 million package in London")
```

```python
text.ents
```

```python
for i in text.ents:
  print(i.text,'-',i.label_,'-',spacy.explain(i.label_))
```
